```python
import numpy as np

# Precios sin impuestos
precios = np.array([10.0, 25.5, 8.0, 50.0])

# NumPy calcula el IVA para TODOS los productos al mismo tiempo
iva = precios * 0.15

# Sumamos los dos arreglos de un solo golpe
precio_total = precios + iva

print("Precios originales:", precios)
print("IVA de cada producto:", iva)
print("Total a pagar:", precio_total)
```

```python
import numpy as np

# Estaturas en metros
estaturas_m = np.array([1.75, 1.82, 1.60, 1.68])

# Multiplicamos todo por 100 de una
estaturas_cm = estaturas_m * 100

# Filtramos rápido a los que midan más de 170 cm
altos = estaturas_cm[estaturas_cm > 170]

print("Estatura en centímetros:", estaturas_cm)
print("Estatura de los que miden más de 170cm:", altos)
```
